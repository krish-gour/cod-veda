```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np 
```

```python
irs = pd.read_csv("1) iris.csv")
```

```python
#Mean
irs[["sepal_length", "sepal_width", "petal_length", "petal_width"]].mean()
```

```python
#Median
irs[["sepal_length", "sepal_width", "petal_length", "petal_width"]].median()
```

```python
#Mode
irs[["sepal_length", "sepal_width", "petal_length", "petal_width"]].mode()
```

```python
#Standard deviation 
irs[["sepal_length", "sepal_width", "petal_length", "petal_width"]].std()
```

```python
#----------------HISTORGRAM#----------------#
```

```python
#length

plt.hist([irs["sepal_length"], irs["petal_length"]],bins=10, alpha=0.7, label=["Sepal Length", "Petal Length"],edgecolor="black", rwidth=0.8)
plt.xlabel("Length")
plt.ylabel("Frequency")
plt.legend()
plt.title("SEPAL & PETAL LENGTH")
plt.show()
```

```python
#Width

plt.hist([irs["sepal_width"], irs["petal_length"]], bins=10, alpha=0.7, label=["Sepal Width", "Petal Width"], edgecolor="black", rwidth=0.8)
plt.xlabel("Width")
plt.ylabel("Frequency")
plt.legend()
plt.title("SEPAL & PETAL WIDTH")
plt.show()
```

```python
#----------------BOXPLOT----------------#
```

```python
#LENGTH

plt.boxplot([irs["petal_length"], irs["sepal_length"]], label=["Petal Length", "Sepal Length"])
plt.grid(True)
plt.xlabel("Measurement Type ")
plt.ylabel("Length(cm)")
plt.title("SEPAL & PETAL Length BOX PLOT")
plt.legend()
plt.show()
```

```python
#WIDTH 

plt.boxplot([irs["petal_width"], irs["sepal_width"]], label=["Petal Width", "Sepal Width"])
plt.grid(True)
plt.xlabel("Measurement Type ")
plt.ylabel("Length(cm)")
plt.title("SEPAL & PETAL Width BOX PLOT")
plt.legend()
plt.show()
```

```python
#----------------SCATTERPLOT----------------#
```

```python
#Petals

plt.scatter(irs.index, irs["petal_length"], label="Petal Length",color="blue", s=15, alpha=0.5)
plt.scatter(irs.index, irs["petal_width"], label="Petal Width",color="orange", s=15, alpha=0.5)
plt.grid(True)
plt.xlabel("Flower Index")
plt.ylabel("Measurement (cm)")
plt.legend()
plt.title("Petal Length and Width")
plt.show()
```

```python
#Sepals 

plt.scatter(irs.index, irs["sepal_length"], label="Sepal Length",edgecolor="black", color="blue", s=15, alpha=0.5)
plt.scatter(irs.index, irs["sepal_width"], label="Sepal Width", edgecolor="black",  color="orange", s=15, alpha=0.5)
plt.grid(True)
plt.xlabel("Flower Index")
plt.ylabel("Measurement (cm)")
plt.legend()
plt.title("Sepal Length and width")
plt.show()
```

```python
#Co-relation
irs[["sepal_length", "sepal_width", "petal_length", "petal_width"]].corr()
```
