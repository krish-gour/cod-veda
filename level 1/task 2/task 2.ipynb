# Exploratory Data Analysis — Iris Dataset

This notebook calculates descriptive statistics and explores the Iris measurements using histograms, boxplots, scatterplots and correlations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

iris = pd.read_csv("1) iris.csv")
features = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
display(iris.head())
print("Dataset shape:", iris.shape)
```

## Descriptive statistics

```python
display(iris[features].describe().T)
display(iris[features].agg(["mean", "median", "std"]).T)
display(iris[features].mode())
```

## Histograms

The width comparison uses **sepal_width** and **petal_width**. The previous version accidentally plotted petal length in the width chart.

```python
plt.figure(figsize=(8, 5))
plt.hist([iris["sepal_length"], iris["petal_length"]], bins=12, alpha=0.7,
         label=["Sepal length", "Petal length"], edgecolor="black")
plt.xlabel("Length (cm)")
plt.ylabel("Frequency")
plt.title("Sepal and Petal Length Distributions")
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.hist([iris["sepal_width"], iris["petal_width"]], bins=12, alpha=0.7,
         label=["Sepal width", "Petal width"], edgecolor="black")
plt.xlabel("Width (cm)")
plt.ylabel("Frequency")
plt.title("Sepal and Petal Width Distributions")
plt.legend()
plt.tight_layout()
plt.show()
```

## Boxplots

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 5))
sns.boxplot(data=iris[["sepal_length", "petal_length"]], ax=axes[0])
axes[0].set_title("Length Measurements")
axes[0].set_ylabel("Length (cm)")
sns.boxplot(data=iris[["sepal_width", "petal_width"]], ax=axes[1])
axes[1].set_title("Width Measurements")
axes[1].set_ylabel("Width (cm)")
plt.tight_layout()
plt.show()
```

## Scatterplots by species

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species", s=55)
plt.title("Petal Length vs Petal Width by Species")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species", s=55)
plt.title("Sepal Length vs Sepal Width by Species")
plt.tight_layout()
plt.show()
```

## Correlation

```python
correlation = iris[features].corr()
display(correlation)
plt.figure(figsize=(7, 5))
sns.heatmap(correlation, annot=True, cmap="coolwarm", center=0, fmt=".2f")
plt.title("Correlation Between Iris Measurements")
plt.tight_layout()
plt.show()
```
