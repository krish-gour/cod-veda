# Stock Price Data Visualization — AAPL

This notebook filters the supplied multi-symbol stock dataset to Apple (AAPL), prepares the date column, and creates line, bar and scatter visualizations.

**Input:** `2) Stock Prices Data Set.csv`  
**Scope:** AAPL observations only

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_FILE = "2) Stock Prices Data Set.csv"
stock = pd.read_csv(DATA_FILE)
required = {"symbol", "date", "open", "high", "low", "close", "volume"}
missing = required.difference(stock.columns)
if missing:
    raise ValueError(f"Dataset is missing expected columns: {sorted(missing)}")

stock["date"] = pd.to_datetime(stock["date"], errors="coerce")
aapl = stock.loc[stock["symbol"].eq("AAPL")].dropna(subset=["date", "close"]).copy()
aapl = aapl.sort_values("date")
print("AAPL rows:", len(aapl))
print("Date range:", aapl["date"].min().date(), "to", aapl["date"].max().date())
display(aapl.head())
```

## 1. Closing price over time

```python
plt.figure(figsize=(12, 5))
plt.plot(aapl["date"], aapl["close"], linewidth=1)
plt.title("AAPL Daily Closing Price")
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig("Figure_1_AAPL_closing_price.png", dpi=160)
plt.show()
```

## 2. Annual trading volume

Annual volume is the sum of daily reported volume for the available AAPL records in each calendar year.

```python
annual_volume = aapl.assign(year=aapl["date"].dt.year).groupby("year")["volume"].sum()
plt.figure(figsize=(10, 5))
annual_volume.plot(kind="bar")
plt.title("AAPL Annual Trading Volume")
plt.xlabel("Year")
plt.ylabel("Total Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("Figure_2_AAPL_annual_volume.png", dpi=160)
plt.show()
```

## 3. Closing price vs trading volume

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=aapl, x="volume", y="close", alpha=0.45, s=18)
plt.title("AAPL Closing Price vs Daily Trading Volume")
plt.xlabel("Daily Trading Volume")
plt.ylabel("Closing Price (USD)")
plt.tight_layout()
plt.savefig("Figure_3_AAPL_price_vs_volume.png", dpi=160)
plt.show()
```
